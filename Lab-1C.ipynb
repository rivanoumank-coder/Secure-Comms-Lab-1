```python
def hex_to_bytes(ciphers):
    return [bytes.fromhex(c) for c in ciphers]

sb = hex_to_bytes(ciphers)
```

```python
def xor_bytes(a, b):
    return bytes(x ^ y for x, y in zip(a, b))
```

```python
def compare_pair(a, b):
    return xor_bytes(a, b)
```

```python
def compare_all(ciphers):
    comparisons = {}

    for i in range(len(ciphers)):
        for j in range(i + 1, len(ciphers)):
            comparisons[(i, j)] = xor_bytes(ciphers[i], ciphers[j])

    return comparisons
```

```python
def space_candidate(byte):
    candidate = byte ^ 0x20

    if 65 <= candidate <= 90 or 97 <= candidate <= 122:
        return chr(candidate)

    return None
```

```python
def calculate_space_scores(ciphers):
    n = len(ciphers)
    max_length = max(len(c) for c in ciphers)

    scores = [[0] * max_length for _ in range(n)]

    for i in range(n):
        for j in range(i + 1, n):

            result = xor_bytes(ciphers[i], ciphers[j])

            for pos, byte in enumerate(result):

                candidate = space_candidate(byte)

                if candidate is not None:
                    scores[i][pos] += 1
                    scores[j][pos] += 1

    return scores
```

```python
def decrypt_position(ciphers, position, key_byte):
    results = []

    for cipher in ciphers:

        if position < len(cipher):
            plaintext_byte = cipher[position] ^ key_byte

            if 32 <= plaintext_byte <= 126:
                results.append(chr(plaintext_byte))
            else:
                results.append('.')
        else:
            results.append('-')

    return results
```

```python
def decrypt_with_keystream(ciphertext, keystream):
    result = []

    for i, byte in enumerate(ciphertext):

        if keystream[i] is not None:
            value = byte ^ keystream[i]

            if 32 <= value <= 126:
                result.append(chr(value))
            else:
                result.append('?')

        else:
            result.append('_')

    return ''.join(result)
```

```python
def test_space(message_number, position):
    """Hypothesis: message `message_number` (1-based) has a space at `position`.
    Show the evidence; do NOT store anything."""
    cipher   = sb[message_number - 1]
    key_byte = recover_keystream_byte(cipher, position, ' ')
    column   = decrypt_position(sb, position, key_byte)

    print(f"Hypothesis: Message {message_number}, position {position} = ' '")
    print(f"Score     : {scores[message_number-1][position]}")
    print(f"Key byte  : 0x{key_byte:02x}")
    for i, ch in enumerate(column, 1):
        print(f"  Message {i:2}: {ch!r}")
    if position < len(secret_bytes):
        print(f"  Message 13: {chr(secret_bytes[position] ^ key_byte)!r}")
    return key_byte
```

```python
def review_candidates(candidates):

    accepted = []
    rejected = []

    for message_number, position in candidates:

        # Don't retest a position where we already have a key byte
        if keystream[position] is not None:
            print()
            print("=" * 60)
            print(f"Position {position} already has a keystream byte.")
            print("Skipping this candidate.")
            continue

        print()
        print("=" * 60)
        print(f"Testing Message {message_number}, Position {position}")
        print("=" * 60)

        # Test the hypothesis
        key_byte = test_space(message_number, position)

        print()
        answer = input("Add this key byte to the keystream? (y/n): ").strip().lower()

        if answer == "y":

            keystream[position] = key_byte
            accepted.append((message_number, position))

            print()
            print(f"✓ Added K[{position}] = 0x{key_byte:02x}")

            print()
            print("Current Message 13:")
            print(decrypt_with_keystream(secret_bytes, keystream))

        else:

            rejected.append((message_number, position))

            print()
            print("✗ Rejected. Keystream unchanged.")

    return accepted, rejected
```

```python
def guess_secret_char(position, guess):
    """
    Guess a plaintext character in Message 13.

    Derives the corresponding keystream byte, tests it against
    Messages 1-12, displays the evidence, and asks whether to
    store the key byte.
    """

    # Basic checks
    if position >= len(secret_bytes):
        print("Position is outside Message 13.")
        return

    if len(guess) != 1:
        print("Guess must be exactly one character.")
        return

    # Derive possible keystream byte from our plaintext guess
    key_byte = secret_bytes[position] ^ ord(guess)

    print()
    print("=" * 60)
    print(f"Guess: Message 13, position {position} = {guess!r}")
    print(f"Possible key byte: 0x{key_byte:02x}")
    print("=" * 60)

    # Test against Messages 1-12
    column = decrypt_position(sb, position, key_byte)

    plausible = 0
    total = 0

    for i, ch in enumerate(column, 1):

        # '-' means that ciphertext does not reach this position
        if ch == '-':
            print(f"Message {i:2}: {ch!r}")
            continue

        total += 1

        # Count letters, spaces and common punctuation as plausible
        if ch.isalpha() or ch == ' ' or ch in ".,!?;:'\"-()":
            plausible += 1

        print(f"Message {i:2}: {ch!r}")

    print("-" * 60)

    if total > 0:
        percentage = plausible / total * 100

        print(
            f"Plausibility: {plausible}/{total} "
            f"({percentage:.1f}%)"
        )

    print()
    print("Current Message 13:")
    print(decrypt_with_keystream(secret_bytes, keystream))

    print()

    answer = input(
        f"Store Message 13 position {position} = {guess!r}? (y/n): "
    ).strip().lower()

    if answer == "y":
        keystream[position] = key_byte

        print()
        print(f"✓ Stored K[{position}] = 0x{key_byte:02x}")

        print()
        print("Updated Message 13:")
        print(decrypt_with_keystream(secret_bytes, keystream))

    else:
        print()
        print("✗ Rejected. Keystream unchanged.")
```

```python
def plaintext_quality(column):
    """
    Calculate how English-like a decrypted column looks.

    Letters and spaces are strongest evidence.
    Common punctuation gets partial credit.
    Strange printable characters get no credit.
    '-' from decrypt_position means that ciphertext
    does not reach this position, so it is ignored.
    """

    score = 0
    total = 0

    for ch in column:

        # Ciphertext does not reach this position
        if ch == '-':
            continue

        total += 1

        # Strong evidence
        if ch.isalpha() or ch == ' ':
            score += 1

        # Possible, but weaker evidence
        elif ch in ".,!?;:'\"-()":
            score += 0.25

        # Everything else gets 0 points

    if total == 0:
        return 0

    return (score / total) * 100


def guess_secret_char(position, guess):
    """
    Guess one plaintext character in Message 13.

    1. Derive the possible keystream byte.
    2. Apply it to Messages 1-12.
    3. Calculate an English-likeness score.
    4. Let the user decide whether to store it.
    """

    # ---------- Validate input ----------

    if position < 0 or position >= len(secret_bytes):
        print("Position is outside Message 13.")
        return

    if len(guess) != 1:
        print("Guess must be exactly one character.")
        return

    # ---------- Derive candidate key byte ----------

    key_byte = secret_bytes[position] ^ ord(guess)

    print()
    print("=" * 60)
    print(f"Guess: Message 13, position {position} = {guess!r}")
    print(f"Possible key byte: 0x{key_byte:02x}")
    print("=" * 60)

    # ---------- Test against Messages 1-12 ----------

    column = decrypt_position(sb, position, key_byte)

    for i, ch in enumerate(column, 1):
        print(f"Message {i:2}: {ch!r}")

    # ---------- Calculate stricter score ----------

    percentage = plaintext_quality(column)

    print("-" * 60)
    print(f"English-likeness score: {percentage:.1f}%")

    # Give a rough interpretation
    if percentage >= 95:
        print("Assessment: VERY STRONG")
    elif percentage >= 90:
        print("Assessment: STRONG")
    elif percentage >= 80:
        print("Assessment: POSSIBLE - inspect carefully")
    elif percentage >= 70:
        print("Assessment: WEAK - be cautious")
    else:
        print("Assessment: POOR")

    # ---------- Show current target ----------

    print()
    print("Current Message 13:")
    print(decrypt_with_keystream(secret_bytes, keystream))

    # Tell us if this position already has a key
    if keystream[position] is not None:
        current_char = chr(
            secret_bytes[position] ^ keystream[position]
        )

        print()
        print(
            f"WARNING: position {position} already decrypts "
            f"as {current_char!r}."
        )

        if keystream[position] != key_byte:
            print(
                "Accepting this guess will REPLACE "
                "the existing key byte."
            )

    # ---------- Human decision ----------

    print()

    answer = input(
        f"Store position {position} = {guess!r}? (y/n): "
    ).strip().lower()

    if answer == "y":

        keystream[position] = key_byte

        print()
        print(
            f"✓ Stored K[{position}] = 0x{key_byte:02x}"
        )

        print()
        print("Updated Message 13:")
        print(
            decrypt_with_keystream(
                secret_bytes,
                keystream
            )
        )

    else:

        print()
        print("✗ Rejected. Keystream unchanged.")
```

```python
def test_secret_phrase(start_position, phrase):
    """
    Test a plaintext phrase against Message 13 without changing
    the keystream.

    For every guessed character:
      - derive the corresponding key byte
      - decrypt that position across Messages 1-12
      - calculate its English-likeness score

    Existing known characters are also checked for conflicts.
    """

    print("=" * 65)
    print(f"Testing phrase {phrase!r} starting at position {start_position}")
    print("=" * 65)

    scores_for_phrase = []

    for offset, guess in enumerate(phrase):
        position = start_position + offset

        if position >= len(secret_bytes):
            print(f"Position {position}: outside Message 13")
            break

        key_byte = secret_bytes[position] ^ ord(guess)
        column = decrypt_position(sb, position, key_byte)
        score = plaintext_quality(column)

        # Check whether we already have a value here
        existing = None

        if keystream[position] is not None:
            existing = chr(
                secret_bytes[position] ^ keystream[position]
            )

        conflict = (
            existing is not None
            and existing != guess
        )

        scores_for_phrase.append(score)

        status = ""

        if conflict:
            status = f"  <-- CONFLICT: currently {existing!r}"
        elif existing == guess:
            status = "  ✓ already confirmed"

        print(
            f"Position {position:3}: "
            f"{guess!r}  "
            f"score={score:5.1f}%"
            f"{status}"
        )

    if scores_for_phrase:
        average = sum(scores_for_phrase) / len(scores_for_phrase)

        print("-" * 65)
        print(f"Average score: {average:.1f}%")

    print()
    print("Current target:")
    print(decrypt_with_keystream(secret_bytes, keystream))
```

```python
def show_context(position, guess, radius=5):
    key_byte = secret_bytes[position] ^ ord(guess)

    print(f"\nTesting position {position} = {guess!r}")
    print("=" * 60)

    for i, cipher in enumerate(sb, 1):

        if position >= len(cipher):
            continue

        start = max(0, position - radius)
        end = min(len(cipher), position + radius + 1)

        chars = []

        for p in range(start, end):

            # Use the candidate key at the position we're testing
            if p == position:
                k = key_byte

            # Use already recovered key bytes elsewhere
            elif keystream[p] is not None:
                k = keystream[p]

            else:
                chars.append("_")
                continue

            value = cipher[p] ^ k

            if 32 <= value <= 126:
                chars.append(chr(value))
            else:
                chars.append(".")

        context = "".join(chars)

        print(
            f"Message {i:2}: "
            f"{start:3}-{end-1:3}  {context}"
        )
```
