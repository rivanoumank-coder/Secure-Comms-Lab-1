```python
def is_prime(p):
    if p < 2:
        return False
    for i in range(2, int(p ** 0.5) + 1):
        if p % i == 0:
            return False
    return True

M = int(input("Enter Message: "))
e = int(input("Enter exponent: "))
p = int(input("Enter prime: "))

if is_prime(p):
    result = pow(M, e, p)
    print("Result:", result)
else:
    print("p must be a prime number")
```

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print("GCD:", gcd(num1, num2))
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


primes = []

for number in range(2, 1000):
    if is_prime(number):
        primes.append(number)

print("Prime numbers:", primes)
print("Highest prime number:", max(primes))
```

```python
def sieve_for_primes_to(n):
    size = n // 2
    sieve = [1] * size
    limit = int(n ** 0.5)

    for i in range(1, limit):
        if sieve[i]:
            val = 2 * i + 1
            tmp = ((size - 1) - i) // val
            sieve[i + val::val] = [0] * tmp

    return [2] + [i * 2 + 1 for i, v in enumerate(sieve)
                  if v and i > 0]


for test in [100, 1000, 5000, 10000]:
    primes = sieve_for_primes_to(test)
    print("Up to", test, ":", max(primes))
```
